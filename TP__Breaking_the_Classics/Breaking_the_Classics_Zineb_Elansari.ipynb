## Warm-Up

### Setup: Caesar function

```python
alphabet = "abcdefghijklmnopqrstuvwxyz"

def caesar(text, key):
    result = ""
    for c in text.lower():
        if c in alphabet:
            pos = alphabet.find(c)
            result = result + alphabet[(pos + key) % 26]
        else:
            result = result + c
    return result
```

## Room 1: The Letter Wall

### Question 1: Encrypt with the substitution alphabet

```python
key = "qmjztgfkpwlsboxncryevhiadu"

def encrypt(text):
    result = ""
    for c in text.lower():
        if c in alphabet:
            result = result + key[alphabet.find(c)]
        else:
            result = result + c
    return result

print(encrypt("KEEP THE SECRET SAFE"))
```

### Question 2: Decrypt with the same alphabet

```python
def decrypt(text):
    result = ""
    for c in text.lower():
        if c in key:
            result = result + alphabet[key.find(c)]
        else:
            result = result + c
    return result

print(decrypt("ERVYE DXVR SXFPJ"))
```

### Question 3: Brute force and frequency analysis
**Can we try every key?** No. A substitution key is a permutation of 26 letters, so there are 26! (about 4 x 10^26) keys. This is too many.

**How can an attacker use letter frequencies?** In English some letters are much more common (E, T, A, O...). The most frequent ciphertext letter is probably E, the next one T, and so on. The rest is found with short words (THE, AND), repeated words and context.

Frequency analysis program:

```python
def frequency(text):
    text = text.lower()
    total = 0
    for c in text:
        if c in alphabet:
            total = total + 1
    results = []
    for c in alphabet:
        n = text.count(c)
        if n > 0:
            results.append([n, c])
    results.sort()
    results.reverse()
    for r in results:
        print(r[1], r[0], round(r[0] * 100 / total, 1), "%")
```

### Question 4: Break the intercepted message
First, the letter frequencies of the message:

```python
cipher1 = """EKT YTJRTE BTYYQFT PY KPZZTO PO EKPY NQRQFRQNK.
STEETR GRTCVTOJD ZXTY OXE RTHTQS EKT IKXST LTD,
MVE PE FPHTY Q FXXZ YEQRE.
SXXL GXR JXBBXO IXRZY, RTNTQETZ NQEETROY,
QOZ YKXRE IXRZY.
IKTO DXV RTJXHTR EKT BTYYQFT,
EKT HPFTOTRT LTD KQY GXVR STEETRY."""

frequency(cipher1)
```

T is the most frequent letter, and the word EKT appears often, so EKT = THE (E=t, K=h, T=e). Other letters come from short words and context. We write the mapping (cipher letter -> plain letter) and apply it:

```python
m = {"e": "t", "k": "h", "t": "e", "y": "s", "j": "c", "r": "r",
     "b": "m", "q": "a", "f": "g", "p": "i", "z": "d", "o": "n",
     "n": "p", "s": "l", "g": "f", "c": "q", "v": "u", "x": "o",
     "h": "v", "i": "w", "l": "k", "m": "b", "d": "y"}

result = ""
for c in cipher1.lower():
    if c in m:
        result = result + m[c]
    else:
        result = result + c
print(result)
```

## Room 2: Break the Vigenere Cipher

### Step 1: Split the ciphertext into 4 groups
The key has 4 letters, so letters 0,4,8... use the 1st key letter, letters 1,5,9... the 2nd, and so on. Each group is a simple Caesar cipher.

```python
cipher2 = "VVHWGQUIVYHIRSUPGTWXJSVIPCWIUPHGCIVIVVHPGHWITSDTRSDVUSYITMZLGFHIXSUCVSDQOSPFGFVLQIOHEVHGMSDGJQRPWAQWGSNXJSPSUHIVGEXIPHVCOPRPCBGYUSLXVCUIECYITHKIMSBAJSQXJSPIUGDKGPHGQAHWEZHETSQXGFWLGOFGGGVAQFGIEZLTUS"
cipher2 = cipher2.lower()

groups = ["", "", "", ""]
for i in range(len(cipher2)):
    groups[i % 4] = groups[i % 4] + cipher2[i]

print(groups)
```

### Step 2: Find the key
In each group, the most frequent letter is probably E. The distance from E to that letter is the shift, which gives the key letter.

```python
vkey = ""
for g in groups:
    best = ""
    best_count = 0
    for c in alphabet:
        n = g.count(c)
        if n > best_count:
            best_count = n
            best = c
    shift = (alphabet.find(best) - alphabet.find("e")) % 26
    print("group", best, best_count, "shift", shift, "key letter", alphabet[shift])
    vkey = vkey + alphabet[shift]

print("Key:", vkey)
```

### Step 3: Decrypt the message and find the access word

```python
plain = ""
for i in range(len(cipher2)):
    shift = alphabet.find(vkey[i % 4])
    plain = plain + alphabet[(alphabet.find(cipher2[i]) - shift) % 26]

print(plain)
```
